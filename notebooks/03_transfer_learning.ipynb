# FarmPulse - EfficientNet transfer learning

EfficientNet-B0 utilise des poids ImageNet et remplace le classifieur final par les classes PlantVillage. Le script reproductible est `src.models.train`.

```python
import torch
import torchvision

print("PyTorch:", torch.__version__)
print("Torchvision:", torchvision.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Training will use CPU")
```

```python
from pathlib import Path
import sys, json, argparse
ROOT = Path.cwd()
while not (ROOT / 'src').is_dir() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from src.models.train import train
from src.utils.tracking import mlflow_tracking_uri

args = argparse.Namespace(processed_dir=str(ROOT / 'data' / 'processed'), model_dir=str(ROOT / 'models'), image_size=224, batch_size=32, epochs=10, learning_rate=3e-4, weight_decay=1e-4, dropout=0.3, num_workers=0, seed=42, pretrained=True, mlflow_uri=mlflow_tracking_uri(ROOT))
metrics = train(args)
metrics
```

Le checkpoint `models/best.pt` contient les poids, les noms de classes et la taille d'entrée. Il est ensuite évalué sur le test set, jamais utilisé pour choisir les hyperparamètres.

```python
import json

print("EfficientNet-B0 training completed!")
print("Validation metrics:", metrics)

metrics_path = ROOT / "models" / "metrics.json"

if metrics_path.exists():
    print(json.loads(metrics_path.read_text()))
else:
    print("Metrics file not found!")
```
